```python
# doing the same shi as Value class, but now in pytorch:

import torch

# We need to define our Value objects in pytorch
# However, Value is based on scalars, but in pytorch,
# we always deal with tensors (N dimensional array of scalars)

# That's why in order to create a scalar value tensor, we 
# just pass it one input in an array:
x1 = torch.Tensor([2.0]).double()
# note: the double is to make everything consistent with python as 
# x1 would otherwise be float32 if not specify .double()
print()
print("non double datatype", torch.Tensor([2.0]).dtype)
print("double datatype", x1.dtype)
print("---")


x2 = torch.Tensor([0.0]).double()
w1 = torch.Tensor([-3.0]).double()
w2 = torch.Tensor([1.0]).double()
b = torch.Tensor([6.8813735870195432]).double()

# by default, pytorch assumes that the above are leaf nodes, hence
# not requiring gradients. That's not true, so we change that property
x1.requires_grad = True
x2.requires_grad = True
w1.requires_grad = True
w2.requires_grad = True
b.requires_grad = True
# (set to False by defualt for efficiency in common cases)

n = x1 * w1 + x2 * w2 + b # just works, like in micrograd
o = torch.tanh(n)

print("o data", o.data.item()) # need to call .item() to strip out the tensor and 
# just return the single element in the tensor

o.backward() # do a backward pass

print("---")
print("x1", x1.data.item())
print("x2", x2.data.item())
print("w1", w1.data.item())
print("w2", w2.data.item())
```
