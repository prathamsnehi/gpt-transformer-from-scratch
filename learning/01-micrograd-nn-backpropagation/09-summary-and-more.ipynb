Neural Nets:
- mathematical expressions
- take input (as data as well as weights and biases)
- do forward pass to get a loss function
- loss function measures the accuracy of the predictions
- low loss means a better model
- use backpropagation to get the gradient
- use the gradient to tune the values of params to lower the loss
- we repeat the step to tune params multiple times (called gradient descent)
- you can make this network do arbitrary things by providing it training data

Real World NN:
- can make it do extremely complex tasks with billions of parameters (ours only had 41 params)
- in GPT, it takes a few words and tries to predict the next word
- when trained on THE ENTIRE INTERNET (100s of billions of params), gets really good at prediction of the right answers
- different ways to calculate loss and perform updates to gradients and the way learning rate is handled that is much more efficient at predicting different data
